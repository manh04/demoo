```python
# Đọc file chứa thông tin tác giả bị thiếu
missing_authors_data = pd.read_excel('authors_data.xlsx')

# In cột của file chứa thông tin tác giả
print("Cột trong missing_authors_data:", missing_authors_data.columns)

# Đảm bảo rằng tên cột trong missing_authors_data là 'title' và 'authors'
missing_authors_data.columns = missing_authors_data.columns.str.strip()  # Xóa khoảng trắng nếu có

# Ghép nối hai bảng dựa trên cột Title
# Sử dụng left join để giữ nguyên tất cả các hàng trong merged_data_unique
book_data = pd.merge(book_data, missing_authors_data, on='title', how='left', suffixes=('', '_from_missing'))

# In cột sau khi ghép nối
print("Cột trong merged_data:", book_data.columns)

# Điền vào các ô authors bị trống bằng thông tin từ file chứa tác giả
if 'authors_from_missing' in book_data.columns:
    book_data['authors'] = book_data['authors'].fillna(book_data['authors_from_missing'])

    # Xóa cột phụ sau khi đã điền dữ liệu
    book_data = book_data.drop(columns=['authors_from_missing'])
else:
    print("Cột 'authors_from_missing' không tồn tại trong book_data.")
```

```python

# Kiểm tra dữ liệu thiếu trong cột quantity
missing_quantity = book_data[book_data['quantity'].isna()]

if not missing_quantity.empty:
    print("Các sách bị thiếu thông tin quantity:")
    print(missing_quantity[['product_id', 'title']])
else:
    print("Không có dữ liệu thiếu trong cột quantity.")

print(len(missing_quantity))
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import LabelEncoder

book_data['quantity'] = pd.to_numeric(book_data['quantity'], errors='coerce')

# Bước 1: Điền dựa trên trung bình theo category
category_mean = book_data.groupby('category')['quantity'].transform('mean')
book_data['quantity'] = book_data['quantity'].fillna(category_mean)

# Bước 2: Điền dựa trên tỷ lệ giữa n_review và quantity
review_ratio = book_data['quantity'].sum() / book_data['n_review'].sum()
book_data.loc[book_data['quantity'].isna(), 'quantity'] = (
    book_data['n_review'] * review_ratio
)

print(book_data['quantity'])
```

```python
# Cắt ngưỡng giá quá cao
merged_data = merged_data[merged_data['original_price'] <= 1_000_000]
import numpy as np

# Tạo biến mới đã log
merged_data['log_current_price'] = np.log1p(merged_data['current_price'])
# Log transform
merged_data['log_quantity'] = np.log1p(merged_data['quantity'])

# Áp dụng IQR trên dữ liệu đã log
Q1 = merged_data['log_quantity'].quantile(0.25)
Q3 = merged_data['log_quantity'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Giữ lại giá trị trong khoảng hợp lý
merged_data = merged_data[(merged_data['log_quantity'] >= lower_bound) &
                          (merged_data['log_quantity'] <= upper_bound)]
# Rule-based: loại bỏ sách có số trang bất hợp lý (>2000 trang)
merged_data = merged_data[merged_data['pages'] <= 2000]

# Sau đó dùng IQR để xử lý các giá trị còn lại
Q1 = merged_data['pages'].quantile(0.25)
Q3 = merged_data['pages'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

merged_data = merged_data[(merged_data['pages'] >= lower_bound) &
                          (merged_data['pages'] <= upper_bound)]
```

### **3.2. Phân tích cảm xúc với PhoBERT**

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
from underthesea import sent_tokenize
import pandas as pd

# Tải mô hình PhoBERT đã được fine-tune cho sentiment analysis
model_name = "wonrax/phobert-base-vietnamese-sentiment"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# Hàm phân tích cảm xúc
def analyze_sentiment(comment):
    # Check if comment is a string and not NaN
    if isinstance(comment, str) and comment != 'nan':
        inputs = tokenizer(comment, return_tensors="pt", truncation=True, padding=True, max_length=256)
        with torch.no_grad():  # Tắt gradient để tăng tốc độ
            outputs = model(**inputs)
        probs = torch.nn.functional.softmax(outputs.logits, dim=-1)
        sentiment = torch.argmax(probs, dim=1).item()
        labels = {0: "Tiêu cực", 1: "Trung lập", 2: "Tích cực"}
        return labels[sentiment]
    else:
        return "Trung lập"  # or any other default value you prefer

# Đọc dữ liệu từ file
data = pd.read_excel('merged_data18124.xlsx')
# Create a new 'sentiment' column if it doesn't exist
if 'sentiment' not in data.columns:
    data['sentiment'] = '' # Initialize with empty strings

# Apply sentiment analysis to the 'content' column instead of the non-existent 'sentiment' column
data['sentiment'] = data['comment'].apply(analyze_sentiment)

# Lưu kết quả ra file mới
data.to_excel('comments_with_sentiment.xlsx', index=False)
print("Đã phân tích cảm xúc xong! Kết quả lưu tại 'comments_with_sentiment.xlsx'")
```

1. Tách bình luận thành các phần: Dùng từ khóa hoặc ngữ pháp để chia bình luận thành phần nói về sách và phần thể hiện cảm xúc tổng thể.
2. Phân tích cảm xúc từng phần: Sử dụng mô hình như PhoBERT để xác định cảm xúc riêng cho từng đoạn.
3. Kết hợp với đánh giá sao: Nếu sao cao (4-5), cảm xúc tổng thể được coi là tích cực, bất kể một phần nội dung có tiêu cực.

```python
import re
from transformers import AutoModelForSequenceClassification, AutoTokenizer
import torch

# Tải mô hình PhoBERT
model_name = "wonrax/phobert-base-vietnamese-sentiment"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# Hàm phân tích cảm xúc
def analyze_sentiment(text):
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=256)
    with torch.no_grad():
        outputs = model(**inputs)
    probs = torch.nn.functional.softmax(outputs.logits, dim=-1)
    sentiment = torch.argmax(probs, dim=1).item()
    labels = {0: "Tiêu cực", 1: "Trung lập", 2: "Tích cực"}
    return labels[sentiment]

# Hàm tách bình luận
def split_comment(comment):
    delivery_keywords = [
    'giao hàng', 'vận chuyển', 'ship', 'nhận hàng', 'đóng gói',
    'chậm', 'nhanh', 'hàng hóa', 'gói hàng', 'shipper', 'vận đơn',
    'shiper', 'gửi', 'nhân viên giao hàng', 'gói', 'tiki', 'dịch vụ',
    'giao', 'shop', 'đóng hàng',
    'bọc',
]
    book_keywords = [
    'sách', 'tác giả', 'nội dung', 'bìa', 'đọc', 'trang', 'chương',
    'cuốn sách', 'cốt truyện', 'nhân vật', 'tựa sách', 'lời tựa',
    'kể', 'câu chuyện', 'câu truyện', 'nội dung', 'tự sự', 'truyện',
    'tác phẩm', 'cuốn', 'lời văn', 'tập', 'quyển', 'văn phong', 'sản phẩm',"trải nghiệm"

]
    overall_keywords = ["thích", "yêu", "tốt", "hài lòng","tuyệt vời","hay","thú vị","xuất sắc"]
    book_quality_keywords = [
    'bìa', 'giấy', 'chất lượng', 'đóng gáy', 'in', 'trang',
    'mực in', 'đóng sách', 'ép', 'phai màu', 'nứt', 'rách',
    'độ dày', 'đóng bìa', 'ép nhựa', 'dày', 'mỏng', 'xốp',
    'bền', 'dính', 'gãy', 'lệch', 'cấn', 'móp', 'méo', 'xước',
    'trầy', 'sờn', 'mờ', 'không sắc nét', 'khổ nhỏ', 'viền',
    'mùi', 'sách cũ', 'màu giấy', 'trắng', 'khổ giấy',
    'vàng', 'nát', 'xé', 'dơ', 'bẩn', 'rạn', 'lỗi in', 'ép lụa',
    'mất trang', 'in thiếu', 'in mờ', 'không đồng đều', 'seal'
]

    sentences = re.split(r'[.,!]', comment)  # Tách theo dấu câu
    book_part = []
    overall_part = []
    delivery_part = []
    book_quality_part = []

    for sentence in sentences:
        sentence = sentence.strip()
        if any(keyword in sentence.lower() for keyword in book_keywords):
            book_part.append(sentence)
        elif any(keyword in sentence.lower() for keyword in overall_keywords):
            overall_part.append(sentence)
        elif any(keyword in sentence.lower() for keyword in delivery_keywords):
           delivery_part.append(sentence)
        elif any(keyword in sentence.lower() for keyword in book_quality_keywords):
            book_quality_part.append(sentence)

    return " ".join(book_part), " ".join(overall_part)

# Ví dụ
comment = merged_data['comment'][3]
rating = merged_data['rating'][3]

# Tách bình luận
book_part, overall_part = split_comment(comment)

# Phân tích cảm xúc
book_sentiment = analyze_sentiment(book_part) if book_part else "Không có"
overall_sentiment = analyze_sentiment(overall_part) if overall_part else "Không có"

# Xác định cảm xúc tổng thể dựa trên sao
overall_rating_sentiment = "Tích cực" if rating >= 4 else ("Tiêu cực" if rating <= 2 else "Trung lập")

# In kết quả
print(f"Bình luận: {comment}")
print(f"Phần về sách: {book_part} → {book_sentiment}")
print(f"Phần tổng thể: {overall_part} → {overall_sentiment}")
print(f"Đánh giá sao: {rating} → Tổng thể: {overall_rating_sentiment}")
print(f"Ghi chú: Nội dung sách cần cải thiện vì {book_part}")
```

### **3.3. Trích xuất chủ đề với BERTopic**

```python
from bertopic import BERTopic
import pandas as pd

# Dữ liệu giả lập
comments = merged_data['comment']
# Khởi tạo và huấn luyện BERTopic
# Giảm số chiều xuống 2 để tránh lỗi
topic_model = BERTopic(language="multilingual", verbose=True,n_gram_range=(1, 2),min_topic_size=2,embedding_model="xlm-roberta-base",low_memory=True,calculate_probabilities=False, nr_topics=2)
# nr_topics='auto' để BERTopic tự động xác định số lượng chủ đề tối ưu


topics, probabilities = topic_model.fit_transform(comments)

# Hiển thị chủ đề
topic_info = topic_model.get_topic_info()
print("Các chủ đề được trích xuất:")
print(topic_info)

# Sinh câu hỏi cải thiện
def generate_improvement_question(topic_keywords):
    main_keyword = topic_keywords[0][0]
    return f"Làm thế nào để cải thiện {main_keyword}?"

for topic in topic_info["Topic"]:
    if topic != -1:
        keywords = topic_model.get_topic(topic)
        question = generate_improvement_question(keywords)
        print(f"Chủ đề {topic}: {keywords} -> {question}")
merged_data.to_excel("comment_with_topic.xlsx",index=False)
```

d. Xử lý tình huống phức tạp

```python
def analyze_complex_comment(comment, rating):
    sentiment_results = analyze_sentiment(comment)

    content_sentiment = [r["sentiment"] for r in sentiment_results if "sách" in r["sentence"].lower()]
    overall_sentiment = [r["sentiment"] for r in sentiment_results if "sách" not in r["sentence"].lower()]

    final_sentiment = "Tích cực" if rating >= 4 else "Tiêu cực"

    return {
        "content_sentiment": content_sentiment[0] if content_sentiment else "Trung lập",
        "overall_sentiment": overall_sentiment[0] if overall_sentiment else final_sentiment,
        "rating": rating
    }

# Ví dụ
comment = "Sách hơi khó hiểu nhưng mình rất thích dịch vụ!"
rating = 4
result = analyze_complex_comment(comment, rating)
print(f"Cảm xúc về nội dung sách: {result['content_sentiment']}")
print(f"Cảm xúc tổng thể: {result['overall_sentiment']}")
print(f"Đánh giá sao: {result['rating']}")
```

  - `Bước 3`: Trực quan hóa kết quả

      - Nhóm các điểm theo từng danh mục sách để so sánh.

**Phân Nhóm Sách**

Phân nhóm (clustering) là một kỹ thuật trong học máy giúp nhóm các đối tượng lại với nhau dựa trên sự tương đồng. Để phân nhóm sách, bạn có thể thực hiện các bước sau:

  - Bước 1: Lựa Chọn Các Đặc Điểm Để Phân Nhóm
  Chọn các biến mà bạn muốn sử dụng cho việc phân nhóm, chẳng hạn như:
    -	Thể loại (category): Phân loại theo thể loại như tiểu thuyết, khoa học, kinh doanh, v.v.
    -	Giá cả (original_price, current_price): Sử dụng giá để xem xét nhóm sách có mức giá tương tự.
    - Số lượng bán (quantity): Để xác định doanh số.
    - 	Điểm đánh giá trung bình (avg_rating): Sử dụng để so sánh chất lượng sách.
  - Bước 2: Chuẩn Bị Dữ Liệu
    -	Chuyển đổi các biến phân loại thành dạng số (one-hot encoding) nếu cần thiết.
    -	Tiến hành chuẩn hóa hoặc chuẩn hóa dữ liệu (scaling) để đảm bảo các đặc điểm có đơn vị khác nhau không ảnh hưởng đến kết quả phân nhóm.
  - Bước 3: Áp Dụng Kỹ Thuật Phân Nhóm
  Sử dụng các thuật toán phân nhóm như K-means, Hierarchical Clustering, hoặc DBSCAN:
    - K-means: Chọn số lượng nhóm (k) trước và thuật toán sẽ phân nhóm các sách thành k nhóm.
    -	Hierarchical Clustering: Tạo ra cây phân cấp để xác định cách các nhóm sách liên kết với nhau.


```python
# Import các thư viện cần thiết
import pandas as pd
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import matplotlib.pyplot as plt
import seaborn as sns

# Tạo DataFrame ví dụ, bổ sung thêm thông tin
data = {
    'category': ['tiểu thuyết', 'khoa học', 'kinh doanh', 'truyện ngắn', 'truyện dài'],
    'original_price': [100, 200, 150, 120, 180],
    'current_price': [90, 180, 140, 110, 170],
    'quantity': [500, 300, 450, 600, 320],
    'avg_rating': [4.5, 4.0, 4.7, 4.6, 4.1],
    'year_published': [2019, 2018, 2020, 2021, 2017],
    'pages': [300, 500, 400, 150, 350]
}
df = pd.DataFrame(data)

# Tiền xử lý dữ liệu với các cột phân loại và số liệu khác
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(), ['category']),
        ('num', StandardScaler(), ['original_price', 'current_price', 'quantity', 'avg_rating', 'year_published', 'pages'])
    ])

# Khởi tạo KMeans và pipeline
kmeans = KMeans(n_clusters=3, random_state=42)
pipeline = Pipeline(steps=[('preprocessor', preprocessor), ('kmeans', kmeans)])

# Huấn luyện mô hình
pipeline.fit(df)
labels = pipeline.predict(df)
df['Cluster'] = labels

# Đánh giá phân nhóm bằng silhouette score
score = silhouette_score(pipeline.named_steps['preprocessor'].transform(df), labels)
print(f'Silhouette Score: {score}')
print(df)

# Sử dụng PCA để giảm chiều và trực quan hóa
pca = PCA(n_components=2)
pca_result = pca.fit_transform(pipeline.named_steps['preprocessor'].transform(df))
df['PCA1'] = pca_result[:, 0]
df['PCA2'] = pca_result[:, 1]
```

#### **4.1.1 Đánh giá khả năng dự đoán của mô hình có sẵn**

```python
import pandas as pd
import time
import torch
from sklearn.metrics import accuracy_score, f1_score
from transformers import AutoTokenizer, AutoModelForSequenceClassification

# Load mô hình
checkpoint = "mr4/phobert-base-vi-sentiment-analysis"
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
model = AutoModelForSequenceClassification.from_pretrained(checkpoint)
model.eval()
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

# Tải dữ liệu đã có cột 'comment' và 'sentiment'
df = pd.read_excel("merged_data_update.xlsx")  # hoặc tên khác nếu bạn có
df = df.dropna(subset=["comment", "sentiment"])

# Map nhãn từ chuỗi -> số (tùy mô hình)
sentiment_map = {"tích cực": 0, "trung lập": 1, "tiêu cực": 2}
true_sentiments = []
pred_sentiments = []

start = time.time()

for _, row in df.iterrows():
    text = row["comment"]
    true = sentiment_map.get(row["sentiment"].lower())
    if true is None:
        continue

    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True, max_length=256).to(device)
    with torch.no_grad():
        outputs = model(**inputs)
        pred = torch.argmax(outputs.logits, dim=1).item()

    true_sentiments.append(true)
    pred_sentiments.append(pred)

end = time.time()

# Tính chỉ số
acc = accuracy_score(true_sentiments, pred_sentiments)
f1 = f1_score(true_sentiments, pred_sentiments, average="weighted")
runtime = end - start
sps = len(true_sentiments) / runtime

# In kết quả
print(f"eval_accuracy: {acc:.4f}")
print(f"eval_f1: {f1:.4f}")
print(f"eval_runtime: {runtime:.2f} giây")
print(f"eval_samples_per_second: {sps:.2f}")
print(f"epoch: 1")
```

```python
import os
import streamlit as st
import pandas as pd
import torch
from transformers import AutoModelForSequenceClassification, AutoTokenizer
from bertopic import BERTopic
from sentence_transformers import SentenceTransformer
import plotly.express as px
from fuzzywuzzy import fuzz
import pickle
import logging
from collections import Counter
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score
import numpy as np
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Table, TableStyle, Paragraph
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib import colors

# ========== SETUP LOGGING ==========
logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")

# ========== SETUP STREAMLIT ==========
st.set_page_config(page_title="Phân tích phản hồi khách hàng", layout="wide")
st.title("📊 Action Insights Dashboard")

# ========== LOAD MODELS ==========
@st.cache_resource
def load_all_models():
    logging.info("Bắt đầu tải các mô hình...")
    try:
        # Tải PhoBERT
        phobert_model = AutoModelForSequenceClassification.from_pretrained("models/fine_tuned_phobert")
        phobert_tokenizer = AutoTokenizer.from_pretrained("models/fine_tuned_phobert")

        # Tải BERTopic
        bertopic_model = BERTopic.load("bertopic_vietnamese_model")

        # Tải SentenceTransformer
        local_model_path = "models/sup-SimCSE-VietNamese-phobert-base"
        if os.path.exists(local_model_path):
            logging.info(f"Tải mô hình SentenceTransformer từ: {local_model_path}")
            embedding_model = SentenceTransformer(local_model_path)
        else:
            logging.info("Tải mô hình SentenceTransformer từ Hugging Face...")
            embedding_model = SentenceTransformer("VoVanPhuc/sup-SimCSE-VietNamese-phobert-base")
            embedding_model.save(local_model_path)
            logging.info(f"Đã lưu mô hình vào: {local_model_path}")

        # Chuyển mô hình sang GPU nếu có
        device = "cuda" if torch.cuda.is_available() else "cpu"
        embedding_model = embedding_model.to(device)

        logging.info("Tải tất cả mô hình thành công!")
        return phobert_model, phobert_tokenizer, bertopic_model, embedding_model
    except Exception as e:
        logging.error(f"Lỗi khi tải mô hình: {e}")
        st.error(f"Lỗi tải mô hình: {e}")
        return None, None, None, None

# ========== LOAD DATA ==========
@st.cache_data
def load_data():
    logging.info("Bắt đầu tải dữ liệu...")
    processed_file = "data/processed_data_all.pkl"
    if os.path.exists(processed_file):
        try:
            df = pd.read_pickle(processed_file)
            logging.info("Tải dữ liệu đã xử lý từ: processed_data_all.pkl")
            return df
        except Exception as e:
            logging.error(f"Lỗi khi đọc file pickle: {e}")
            st.warning("Lỗi đọc file đã xử lý, tải file gốc...")

    try:
        df = pd.read_excel("data/merged_data_update.xlsx")
        required_columns = ["comment", "title", "rating"]
        missing_cols = [col for col in required_columns if col not in df.columns]
        if missing_cols:
            logging.error(f"Thiếu cột: {missing_cols}")
            st.error(f"Thiếu cột: {missing_cols}")
            return pd.DataFrame()
        df["comment"] = df["comment"].astype(str).fillna("")
        logging.info("Tải dữ liệu thành công từ merged_data_update.xlsx")
        return df
    except FileNotFoundError:
        logging.error("Không tìm thấy file merged_data_update.xlsx")
        st.error("Không tìm thấy file merged_data_update.xlsx!")
        return pd.DataFrame()

# ========== COMPUTE EMBEDDINGS ==========
@st.cache_data
@st.cache_data
def compute_embeddings(_comments, _embedding_model):
    cache_file = "data/embeddings_cache.pkl"
    comments = pd.Series(_comments)

    if os.path.exists(cache_file):
        try:
            with open(cache_file, "rb") as f:
                cached_data = pickle.load(f)
                if cached_data["comments"].equals(comments):
                    logging.info("Tải embeddings từ cache")
                    embeddings = np.array(cached_data["embeddings"], dtype=np.float32)  # ✅ Sửa chỗ này
                    return embeddings
        except Exception as e:
            logging.warning(f"Lỗi đọc cache embeddings: {e}")

    logging.info("Tính toán embeddings với batch...")
    batch_size = 32
    embeddings = []
    for i in range(0, len(comments), batch_size):
        batch_comments = comments[i:i + batch_size].tolist()
        try:
            batch_embeddings = _embedding_model.encode(
                batch_comments,
                batch_size=batch_size,
                show_progress_bar=True,
                device="cuda" if torch.cuda.is_available() else "cpu"
            )
            embeddings.extend(batch_embeddings)
        except Exception as e:
            logging.error(f"Lỗi khi mã hóa batch {i//batch_size}: {e}")
            embeddings.extend([[] for _ in range(len(batch_comments))])

    try:
        with open(cache_file, "wb") as f:
            pickle.dump({"comments": comments, "embeddings": embeddings}, f)
        logging.info("Đã lưu embeddings vào cache")
    except Exception as e:
        logging.warning(f"Lỗi lưu cache embeddings: {e}")

    return np.array(embeddings, dtype=np.float32)  # ✅ Sửa chỗ này

# ========== TRUST SCORE FUNCTIONS ==========
@st.cache_resource
def load_phobert():
    try:
        model = AutoModelForSequenceClassification.from_pretrained("models/fine_tuned_phobert")
        tokenizer = AutoTokenizer.from_pretrained("models/fine_tuned_phobert")
        device = "cuda" if torch.cuda.is_available() else "cpu"
        return model.to(device), tokenizer
    except Exception as e:
        logging.error(f"Lỗi tải PhoBERT: {e}")
        st.error(f"Lỗi tải PhoBERT: {e}")
        return None, None

phobert_model, phobert_tokenizer = load_phobert()

def predict_sentiment(comment):
    if not comment or not phobert_model or not phobert_tokenizer:
        return "neutral", 0.0
    try:
        inputs = phobert_tokenizer(comment, return_tensors="pt", truncation=True, padding=True, max_length=256)
        inputs = {k: v.to(phobert_model.device) for k, v in inputs.items()}
        with torch.no_grad():
            outputs = phobert_model(**inputs)
        logits = outputs.logits
        probs = torch.softmax(logits, dim=1)
        predicted_class = torch.argmax(logits, dim=1).item()
        max_prob = torch.max(probs, dim=1).values.item()
        sentiment_map = {0: "negative", 1: "neutral", 2: "positive"}
        return sentiment_map[predicted_class], max_prob
    except Exception as e:
        logging.error(f"Lỗi dự đoán sentiment: {e}")
        return "neutral", 0.0

def has_meaningful_content(comment):
    if not comment or not phobert_model or not phobert_tokenizer:
        return False
    try:
        inputs = phobert_tokenizer(comment, return_tensors="pt", truncation=True, padding=True, max_length=256)
        inputs = {k: v.to(phobert_model.device) for k, v in inputs.items()}
        with torch.no_grad():
            outputs = phobert_model(**inputs)
        logits = outputs.logits
        probs = torch.softmax(logits, dim=1)
        max_prob = torch.max(probs, dim=1).values.item()
        return max_prob > 0.65
    except Exception as e:
        logging.error(f"Lỗi kiểm tra nội dung: {e}")
        return False

def check_length(comment):
    words = comment.split()
    length = len(words)
    if length < 10:
        if has_meaningful_content(comment):
            return False, 0.0
        return True, 0.2
    elif length > 200:
        return True, 0.1
    return False, 0.0

def has_repetitive_words(comment):
    words = comment.split()
    if len(words) < 15:
        return False
    unique_ratio = len(set(words)) / len(words)
    word_counts = Counter(words)
    max_repetition = max(word_counts.values())
    return unique_ratio < 0.5 or max_repetition > 5

def has_spam_words(comment):
    spam_words = ['...', 'spam', 'quảng_cáo', 'khuyến_mãi']  # Mở rộng danh sách
    words = comment.lower().split()
    if len(words) <= 5 and any(word in spam_words for word in words):
        return not has_meaningful_content(comment)
    return False

def sentiment_conflict(predicted_sentiment, rating, confidence_threshold=0.85, predicted_confidence=0.0):
    if pd.isna(rating) or predicted_confidence < confidence_threshold:
        return False
    if rating >= 4 and predicted_sentiment == "negative":
        return True
    if rating <= 2 and predicted_sentiment == "positive":
        return True
    return False

def compute_trust(comment, sentiment, rating):
    reasons = []
    scores = []
    comment = comment or ""

    is_length_issue, length_score = check_length(comment)
    if is_length_issue:
        reasons.append("Độ dài không phù hợp")
        scores.append(length_score)

    if has_repetitive_words(comment):
        reasons.append("Lặp từ")
        scores.append(0.2)

    if has_spam_words(comment):
        reasons.append("Từ spam")
        scores.append(0.1)

    predicted_sentiment, predicted_confidence = predict_sentiment(comment)
    if sentiment_conflict(predicted_sentiment, rating, predicted_confidence=predicted_confidence):
        reasons.append("Xung đột sentiment-rating")
        scores.append(0.5)

    total_score = sum(scores)
    label = "Không đáng tin cậy" if total_score > 0.5 else "Đáng tin cậy"
    return label, reasons

# ========== ACTION RECOMMENDATION MAPPING ==========
action_mapping = {
    ("Nội dung sách", "positive"): "Đề xuất quảng bá sách hoặc sản phẩm tương tự.",
    ("Nội dung sách", "negative"): "Thông báo bộ phận nội dung kiểm tra phản ánh.",
    ("Chất lượng sách", "negative"): "Liên hệ nhà in để kiểm tra chất lượng sản phẩm.",
    ("Chất lượng giao hàng", "negative"): "Gửi phản hồi tới bộ phận vận chuyển và đề xuất hỗ trợ khách hàng.",
    ("Giá cả", "neutral"): "Xem xét chương trình khuyến mãi hoặc đối chiếu với đối thủ.",
    ("Không xác định", "negative"): "Gửi khảo sát chi tiết để làm rõ vấn đề từ phản hồi khách hàng.",
}

def get_action_recommendation(topic, sentiment):
    return action_mapping.get((topic, sentiment), "Không có gợi ý cụ thể.")

# ========== PROCESS DATA ==========
def map_topic_to_label(topic_keywords):
    if not topic_keywords:
        return "Không xác định"
    keywords = [kw[0].lower() for kw in topic_keywords[:5]]  # Lấy top 5 từ khóa
    topic_mapping = {
        "sách": "Nội dung sách",
        "giao_hàng": "Chất lượng giao hàng",
        "vận_chuyển": "Chất lượng giao hàng",
        "chất_lượng": "Chất lượng sách",
        "in_ấn": "Chất lượng sách",
        "giá": "Giá cả",
        "dịch_vụ": "Dịch vụ khách hàng",
        "không": "Không liên quan",
        "spam": "Không liên quan"
    }
    for keyword, label in topic_mapping.items():
        for kw in keywords:
            if fuzz.partial_ratio(keyword, kw) > 80:
                return label
    return "Không xác định"

@st.cache_data
def process_data(_df, _topic_model, _embedding_model):
    processed_file = "data/processed_data_all.pkl"
    if os.path.exists(processed_file):
        try:
            df = pd.read_pickle(processed_file)
            logging.info("Tải dữ liệu đã xử lý từ: processed_data_all.pkl")
            return df
        except Exception as e:
            logging.error(f"Lỗi đọc file pickle: {e}")
            st.warning("Lỗi đọc file đã xử lý, xử lý lại dữ liệu...")

    df = _df.copy()
    if df.empty:
        logging.error("Dữ liệu rỗng, không thể xử lý")
        return df

    logging.info("Bắt đầu xử lý dữ liệu...")

    # Tính độ tin cậy
    if "trust_label" not in df.columns or df["trust_label"].isna().all():
        logging.info("Tính toán độ tin cậy cho phản hồi...")
        trust_labels = []
        trust_reasons = []
        for _, row in df.iterrows():
            label, reasons = compute_trust(row["comment"], row.get("sentiment", "neutral"), row.get("rating", np.nan))
            trust_labels.append(label)
            trust_reasons.append(reasons)
        df["trust_label"] = trust_labels
        df["trust_reasons"] = trust_reasons

    # Dự đoán chủ đề
    if "topic_vo" not in df.columns or df["topic_vo"].isna().all():
        if "action_recommendation" not in df.columns or df["action_recommendation"].isna().all():
            logging.info("Tạo cột gợi ý hành động...")
            df["action_recommendation"] = df.apply(
            lambda row: get_action_recommendation(row["topic_vo"], row["sentiment"]),
            axis=1
        )
        logging.info("Dự đoán chủ đề bằng BERTopic...")
        comments = df["comment"].tolist()
        if comments and _topic_model and _embedding_model:
            try:
                embeddings = compute_embeddings(comments, _embedding_model)
                topics, _ = _topic_model.transform(comments, embeddings=embeddings)
                topic_labels = []
                for topic_id in topics:
                    if topic_id == -1:
                        topic_labels.append("Không xác định")
                    else:
                        topic_keywords = _topic_model.get_topic(topic_id)
                        label = map_topic_to_label(topic_keywords)
                        topic_labels.append(label)
                df["topic_vo"] = topic_labels
            except Exception as e:
                logging.error(f"Lỗi khi dự đoán chủ đề: {e}")
                df["topic_vo"] = ["unknown"] * len(df)
        else:
            logging.warning("Không thể dự đoán chủ đề do thiếu mô hình hoặc dữ liệu")
            df["topic_vo"] = ["unknown"] * len(df)

    # Tính độ tương đồng giữa topic_vo và topic_comment
    if 'topic_comment' in df.columns and 'topic_vo' in df.columns:
            # Gợi ý hành động
        logging.info("Tính độ tương đồng giữa topic_vo và topic_comment...")
        valid_idx = df['topic_comment'].notna() & df['topic_vo'].notna()
        true_labels = df.loc[valid_idx, 'topic_comment']
        pred_labels = df.loc[valid_idx, 'topic_vo']

        # Chỉ tính nếu true_labels và pred_labels không rỗng
        if len(true_labels) > 0 and len(pred_labels) > 0:
            try:
                ari_score = adjusted_rand_score(true_labels, pred_labels)
                nmi_score = normalized_mutual_info_score(true_labels, pred_labels)
                accuracy = np.mean(true_labels == pred_labels)

                st.write("✅ Độ tương đồng giữa topic_vo và topic_comment:")
                st.write(f"  - Adjusted Rand Index (ARI): {ari_score:.4f}")
                st.write(f"  - Normalized Mutual Information (NMI): {nmi_score:.4f}")
                st.write(f"  - Tỷ lệ trùng khớp (Accuracy): {accuracy:.4f}")
                logging.info(f"Độ tương đồng - ARI: {ari_score:.4f}, NMI: {nmi_score:.4f}, Accuracy: {accuracy:.4f}")
            except Exception as e:
                logging.error(f"Lỗi tính độ tương đồng: {e}")
                st.warning(f"Lỗi tính độ tương đồng: {e}")

    # Đánh dấu phản hồi tiêu stt:negative_and_trustworthy
    df["negative_and_trustworthy"] = (df["sentiment"] == "negative") & (df["trust_label"] == "Đáng tin cậy")

    # Lưu dữ liệu đã xử lý
    try:
        df.to_pickle(processed_file)
        logging.info("Đã lưu dữ liệu đã xử lý vào: processed_data_all.pkl")
    except Exception as e:
        logging.error(f"Lỗi lưu file pickle: {e}")
        st.warning(f"Lỗi lưu dữ liệu: {e}")

    return df

# ========== SUMMARIZE BOOKS ==========
@st.cache_data
def summarize_books(_df):
    if _df.empty:
        logging.error("Dữ liệu rỗng, không thể tổng hợp")
        return pd.DataFrame()

    id_cols = ["title", "authors", "category", "original_price", "current_price", "manufacturer"]
    try:
        book_summary = _df.groupby(id_cols).agg({
            "rating": ["mean", "count"],
            "sentiment": lambda x: {
                "Tích cực": sum(1 for s in x if s == "positive"),
                "Trung lập": sum(1 for s in x if s == "neutral"),
                "Tiêu cực": sum(1 for s in x if s == "negative")
            },
            "trust_label": lambda x: sum(1 for t in x if t == "Đáng tin cậy"),
            "negative_and_trustworthy": lambda x: sum(1 for t in x if t)
        }).reset_index()
        book_summary.columns = [
            "title", "authors", "category", "original_price", "current_price", "manufacturer",
            "avg_rating", "n_reviews", "sentiment_breakdown", "n_trustworthy", "n_negative_trustworthy"
        ]
        book_summary["avg_rating"] = book_summary["avg_rating"].round(2)
        return book_summary
    except Exception as e:
        logging.error(f"Lỗi khi tổng hợp sách: {e}")
        st.error(f"Lỗi tổng hợp sách: {e}")
        return pd.DataFrame()

# ========== FUZZY SEARCH ==========
@st.cache_data
def fuzzy_search(_df, search_term, threshold=70):
    if _df.empty or not search_term:
        return _df

    try:
        matches = []
        for idx, row in _df.iterrows():
            book_info = f"{row['title']} {row['authors']} {row['category']}".lower()
            score = fuzz.partial_ratio(search_term.lower(), book_info)
            if score >= threshold:
                matches.append((idx, score))
        matches.sort(key=lambda x: x[1], reverse=True)
        matched_indices = [m[0] for m in matches]
        return _df.iloc[matched_indices]
    except Exception as e:
        logging.error(f"Lỗi tìm kiếm fuzzy: {e}")
        st.error(f"Lỗi tìm kiếm: {e}")
        return _df

# ========== BOOK RECOMMENDATION ==========
@st.cache_data
def recommend_books(_book_summary, selected_book, top_n=5):
    if _book_summary.empty or not selected_book:
        return pd.DataFrame()

    try:
        selected = _book_summary[_book_summary["title"] == selected_book].iloc[0]
        category = selected["category"]
        avg_rating = selected["avg_rating"]

        # Lọc sách cùng danh mục, có rating tương tự (±1) và ưu tiên n_trustworthy
        candidates = _book_summary[
            (_book_summary["category"] == category) &
            (_book_summary["title"] != selected_book) &
            (_book_summary["avg_rating"].between(avg_rating - 1, avg_rating + 1))
        ]

        if candidates.empty:
            return pd.DataFrame()

        # Sắp xếp theo n_trustworthy và avg_rating
        recommendations = candidates.sort_values(
            by=["n_trustworthy", "avg_rating"],
            ascending=[False, False]
        ).head(top_n)

        return recommendations
    except Exception as e:
        logging.error(f"Lỗi gợi ý sách: {e}")
        return pd.DataFrame()

# ========== EXPORT REPORT ==========
def export_to_pdf(book_summary):
    try:
        pdf_file = "output/book_summary_report.pdf"
        os.makedirs("output", exist_ok=True)
        doc = SimpleDocTemplate(pdf_file, pagesize=letter)
        styles = getSampleStyleSheet()
        elements = []

        # Tiêu đề
        elements.append(Paragraph("Báo cáo tổng quan sách", styles["Title"]))

        # Dữ liệu bảng
        data = [["Tên sách", "Tác giả", "Thể loại", "Giá hiện tại", "Rating TB", "Số đánh giá", "Đánh giá đáng tin"]]
        for _, row in book_summary.iterrows():
            data.append([
                row["title"],
                row["authors"],
                row["category"],
                f"{row['current_price']} VNĐ",
                f"{row['avg_rating']:.2f}",
                str(row["n_reviews"]),
                str(row["n_trustworthy"])
            ])

        table = Table(data)
        table.setStyle(TableStyle([
            ('BACKGROUND', (0, 0), (-1, 0), colors.grey),
            ('TEXTCOLOR', (0, 0), (-1, 0), colors.whitesmoke),
            ('ALIGN', (0, 0), (-1, -1), 'CENTER'),
            ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold'),
            ('FONTSIZE', (0, 0), (-1, 0), 12),
            ('BOTTOMPADDING', (0, 0), (-1, 0), 12),
            ('BACKGROUND', (0, 1), (-1, -1), colors.beige),
            ('GRID', (0, 0), (-1, -1), 1, colors.black)
        ]))
        elements.append(table)

        doc.build(elements)
        return pdf_file
    except Exception as e:
        logging.error(f"Lỗi xuất PDF: {e}")
        st.error(f"Lỗi xuất PDF: {e}")
        return None

# ========== DASHBOARD ==========
def main():
    # Tải mô hình
    phobert_model, phobert_tokenizer, topic_model, embedding_model = load_all_models()
    if all(x is None for x in [phobert_model, phobert_tokenizer, topic_model, embedding_model]):
        st.error("Không thể tiếp tục do lỗi tải mô hình!")
        return

    # Tải dữ liệu
    if 'df' not in st.session_state:
        raw_data = load_data()
        if raw_data.empty:
            st.error("Không có dữ liệu để hiển thị!")
            return
        st.session_state.df = process_data(raw_data, topic_model, embedding_model)
    df = st.session_state.df

    if df.empty:
        st.error("Dữ liệu rỗng, vui lòng kiểm tra file dữ liệu!")
        return

    # Sidebar: Bộ lọc và xuất báo cáo
    with st.sidebar:
        st.header("📌 Tuỳ chọn")
        search_term = st.text_input("Tìm kiếm sách (tên, tác giả, thể loại)", "")
        categories = st.multiselect(
            "Chọn chủ đề phản hồi",
            options=df["topic_vo"].unique(),
            default=df["topic_vo"].unique()
        )
        sentiments = st.multiselect(
            "Chọn cảm xúc",
            options=df["sentiment"].unique(),
            default=df["sentiment"].unique()
        )
        trust_filter = st.radio("Độ tin cậy phản hồi", ["Tất cả", "Đáng tin cậy", "Không đáng tin cậy"])

        st.markdown("### Xuất báo cáo")
        if st.button("Xuất Excel"):
            book_summary = summarize_books(df)
            if not book_summary.empty:
                os.makedirs("output", exist_ok=True)
                excel_file = "output/book_summary.xlsx"
                book_summary.to_excel(excel_file, index=False)
                st.success(f"Đã xuất báo cáo Excel vào {excel_file}")
            else:
                st.error("Không có dữ liệu để xuất!")

        if st.button("Xuất PDF"):
            book_summary = summarize_books(df)
            if not book_summary.empty:
                pdf_file = export_to_pdf(book_summary)
                if pdf_file:
                    st.success(f"Đã xuất báo cáo PDF vào {pdf_file}")
            else:
                st.error("Không có dữ liệu để xuất!")

    # Tạo tổng quan sách
    book_summary = summarize_books(df)
    if search_term:
        book_summary = fuzzy_search(book_summary, search_term)

    # Tab layout
    tab1, tab2, tab3, tab4 = st.tabs(["📚 Tổng quan sách", "📝 Chi tiết phản hồi", "📖 Gợi ý sách", "📌 Gợi ý hành động"])


    # Tab 1: Tổng quan sách
    with tab1:
        st.markdown(f"### 🔎 Tổng quan: {len(book_summary)} sách")
        if not book_summary.empty:
            books_data = []
            for _, row in book_summary.iterrows():
                books_data.append({
                    "Tên sách": row["title"],
                    "Tác giả": row["authors"],
                    "Thể loại": row["category"],
                    "Giá gốc": row["original_price"],
                    "Giá hiện tại": row["current_price"],
                    "Nhà xuất bản": row["manufacturer"],
                    "Rating trung bình": row["avg_rating"],
                    "Số lượng đánh giá": row["n_reviews"],
                    "Phân khúc cảm xúc": f"Tích cực: {row['sentiment_breakdown']['Tích cực']}, Trung lập: {row['sentiment_breakdown']['Trung lập']}, Tiêu cực: {row['sentiment_breakdown']['Tiêu cực']}",
                    "Đánh giá đáng tin": row["n_trustworthy"],
                    "Phản hồi tiêu cực đáng tin": row["n_negative_trustworthy"]
                })
            books_df = pd.DataFrame(books_data)
            st.dataframe(
                books_df.head(100),
                column_config={
                    "Tên sách": st.column_config.TextColumn("Tên sách", width="medium"),
                    "Tác giả": st.column_config.TextColumn("Tác giả", width="medium"),
                    "Thể loại": st.column_config.TextColumn("Thể loại", width="medium"),
                    "Giá gốc": st.column_config.NumberColumn("Giá gốc (VNĐ)"),
                    "Giá hiện tại": st.column_config.NumberColumn("Giá hiện tại (VNĐ)"),
                    "Nhà xuất bản": st.column_config.TextColumn("Nhà xuất bản", width="medium"),
                    "Rating trung bình": st.column_config.NumberColumn("Rating trung bình", min_value=1, max_value=5),
                    "Số lượng đánh giá": st.column_config.NumberColumn("Số lượng đánh giá"),
                    "Phân khúc cảm xúc": st.column_config.TextColumn("Phân khúc cảm xúc", width="medium"),
                    "Đánh giá đáng tin": st.column_config.NumberColumn("Đánh giá đáng tin"),
                    "Phản hồi tiêu cực đáng tin": st.column_config.NumberColumn("Phản hồi tiêu cực đáng tin")
                },
                height=300
            )
        else:
            st.warning("Không có dữ liệu sách để hiển thị!")

    # Tab 2: Chi tiết phản hồi
    with tab2:
        st.markdown("### 📌 Phân tích chi tiết phản hồi")
        if not book_summary.empty:
            selected_book = st.selectbox("Chọn một cuốn sách", book_summary["title"].tolist())
            if selected_book:
                book_data = book_summary[book_summary["title"] == selected_book].iloc[0]
                book_reviews = df[df["title"] == selected_book]
                filtered_reviews = book_reviews[
                    (book_reviews["topic_vo"].isin(categories)) &
                    (book_reviews["sentiment"].isin(sentiments))
                ]
                if trust_filter != "Tất cả":
                    filtered_reviews = filtered_reviews[filtered_reviews["trust_label"] == trust_filter]

                st.markdown(f"#### 📖 Sách: {selected_book}")
                col1, col2 = st.columns(2)
                with col1:
                    st.write(f"**Tác giả:** {book_data['authors']}")
                    st.write(f"**Thể loại:** {book_data['category']}")
                    st.write(f"**Giá gốc:** {book_data['original_price']} VNĐ")
                    st.write(f"**Giá hiện tại:** {book_data['current_price']} VNĐ")
                    st.write(f"**Nhà xuất bản:** {book_data['manufacturer']}")
                    st.write(f"**Rating trung bình:** {book_data['avg_rating']}/5")
                    st.write(f"**Số lượng đánh giá:** {book_data['n_reviews']}")
                    st.write(f"**Phản hồi tiêu cực đáng tin:** {book_data['n_negative_trustworthy']}")

                with col2:
                    if st.checkbox("Hiển thị biểu đồ phân khúc cảm xúc"):
                        sentiment_data = book_data["sentiment_breakdown"]
                        try:
                            fig1 = px.pie(
                                values=list(sentiment_data.values()),
                                names=list(sentiment_data.keys()),
                                title="Phân khúc cảm xúc",
                                color_discrete_sequence=["#00CC96", "#EF553B", "#FFA15A"]
                            )
                            st.plotly_chart(fig1, use_container_width=True)
                        except Exception as e:
                            st.warning(f"Lỗi hiển thị biểu đồ cảm xúc: {e}")

                    if st.checkbox("Hiển thị biểu đồ phân bố chủ đề"):
                        topic_counts = filtered_reviews["topic_vo"].value_counts()
                        try:
                            fig3 = px.pie(
                                values=topic_counts.values,
                                names=topic_counts.index,
                                title="Phân bố chủ đề phản hồi",
                                color_discrete_sequence=px.colors.qualitative.Plotly
                            )
                            st.plotly_chart(fig3, use_container_width=True)
                        except Exception as e:
                            st.warning(f"Lỗi hiển thị biểu đồ chủ đề: {e}")

                st.markdown("**📝 Tất cả phản hồi:**")
                if not filtered_reviews.empty:
                    reviews_data = filtered_reviews[[
                        "comment", "sentiment", "topic_vo", "rating",
                        "trust_label", "trust_reasons", "negative_and_trustworthy"
                    ]]
                    st.dataframe(
                        reviews_data.head(50),
                        column_config={
                            "comment": st.column_config.TextColumn("Phản hồi", width="large"),
                            "sentiment": st.column_config.TextColumn("Cảm xúc", width="medium"),
                            "topic_vo": st.column_config.TextColumn("Chủ đề", width="medium"),
                            "rating": st.column_config.NumberColumn("Rating", min_value=1, max_value=5),
                            "trust_label": st.column_config.TextColumn("Độ tin cậy", width="medium"),
                            "trust_reasons": st.column_config.TextColumn("Lý do", width="large"),
                            "negative_and_trustworthy": st.column_config.CheckboxColumn(
                                "Tiêu cực & Đáng tin",
                                default=False
                            )
                        },
                        height=300
                    )
                else:
                    st.warning("Không có phản hồi nào phù hợp với bộ lọc!")

                st.markdown("**🔍 Phân tích chi tiết một phản hồi:**")
                if not filtered_reviews.empty:
                    selected_comment = st.selectbox("Chọn một phản hồi để phân tích", filtered_reviews["comment"].tolist())
                    if selected_comment:
                        comment_data = filtered_reviews[filtered_reviews["comment"] == selected_comment].iloc[0]
                        st.write("**Nội dung phản hồi:**")
                        st.write(comment_data["comment"])
                        st.write(f"**Cảm xúc:** {comment_data['sentiment']}")
                        st.write(f"**Chủ đề:** {comment_data['topic_vo']}")
                        st.write(f"**Rating:** {comment_data['rating']}")
                        st.write(f"**Độ tin cậy:** {comment_data['trust_label']}")
                        st.write(f"**Tiêu cực & Đáng tin:** {'Có' if comment_data['negative_and_trustworthy'] else 'Không'}")
                        if comment_data["trust_reasons"]:
                            st.write("**Lý do:**")
                            for reason in comment_data["trust_reasons"]:
                                st.write(f"- {reason}")
                else:
                    st.info("Không có phản hồi để phân tích.")

                if topic_model and st.checkbox("Hiển thị ma trận khoảng cách chủ đề"):
                    try:
                        topic_info = topic_model.get_topic_info()
                        topic_labels = {}
                        topic_data = []
                        for _, row in topic_info.iterrows():
                            topic_id = row['Topic']
                            if topic_id == -1:
                                continue
                            topic_keywords = topic_model.get_topic(topic_id)
                            label = map_topic_to_label(topic_keywords)
                            topic_labels[topic_id] = label
                            topic_data.append({
                                "Chủ đề ID": topic_id,
                                "Tên chủ đề": label,
                                "Số lượng phản hồi": row['Count'],
                                "Từ khóa chính": ", ".join([kw[0] for kw in topic_keywords[:3]])
                            })

                        fig_topics = topic_model.visualize_topics(
                            custom_labels=topic_labels,
                            width=800, height=600
                        )
                        st.plotly_chart(fig_topics, use_container_width=True)

                        st.markdown("**Danh sách các chủ đề:**")
                        topic_df = pd.DataFrame(topic_data)
                        st.dataframe(topic_df, use_container_width=True)
                    except Exception as e:
                        st.warning(f"Lỗi hiển thị ma trận chủ đề: {e}")

    # Tab 3: Gợi ý sách
    with tab3:
        st.markdown("### 📖 Gợi ý sách")
        if not book_summary.empty:
            selected_book = st.selectbox("Chọn sách để nhận gợi ý", book_summary["title"].tolist(), key="recommend_book")
            if selected_book:
                recommendations = recommend_books(book_summary, selected_book)
                if not recommendations.empty:
                    st.markdown(f"#### Gợi ý sách tương tự '{selected_book}':")
                    rec_data = []
                    for _, row in recommendations.iterrows():
                        rec_data.append({
                            "Tên sách": row["title"],
                            "Tác giả": row["authors"],
                            "Thể loại": row["category"],
                            "Giá hiện tại": row["current_price"],
                            "Rating trung bình": row["avg_rating"],
                            "Số đánh giá": row["n_reviews"],
                            "Đánh giá đáng tin": row["n_trustworthy"]
                        })
                    rec_df = pd.DataFrame(rec_data)
                    st.dataframe(
                        rec_df,
                        column_config={
                            "Tên sách": st.column_config.TextColumn("Tên sách", width="medium"),
                            "Tác giả": st.column_config.TextColumn("Tác giả", width="medium"),
                            "Thể loại": st.column_config.TextColumn("Thể loại", width="medium"),
                            "Giá hiện tại": st.column_config.NumberColumn("Giá hiện tại (VNĐ)"),
                            "Rating trung bình": st.column_config.NumberColumn("Rating trung bình", min_value=1, max_value=5),
                            "Số đánh giá": st.column_config.NumberColumn("Số đánh giá"),
                            "Đánh giá đáng tin": st.column_config.NumberColumn("Đánh giá đáng tin")
                        },
                        height=200
                    )
                else:
                    st.warning("Không tìm thấy sách phù hợp để gợi ý!")
        else:
            st.warning("Không có dữ liệu sách để gợi ý!")
    # Tab 4: Gợi ý hành động
    with tab4:
        st.markdown("### 🔧 Danh sách phản hồi có hành động gợi ý")
        action_df = df[df["action_recommendation"] != "Không có gợi ý cụ thể."]

        if not action_df.empty:
            st.dataframe(
                action_df[[
                    "comment", "topic_vo", "sentiment", "trust_label", "action_recommendation"
                ]].head(100),
                column_config={
                    "comment": st.column_config.TextColumn("Phản hồi", width="large"),
                    "topic_vo": st.column_config.TextColumn("Chủ đề", width="medium"),
                    "sentiment": st.column_config.TextColumn("Cảm xúc", width="medium"),
                    "trust_label": st.column_config.TextColumn("Độ tin cậy", width="medium"),
                    "action_recommendation": st.column_config.TextColumn("Gợi ý hành động", width="large")
                },
                height=400
            )
        else:
            st.info("Không có phản hồi nào có gợi ý hành động.")


# ========== RUN ==========
if __name__ == "__main__":
    main()
```
